```python
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
import matplotlib.pyplot as plt

sns.set_style("darkgrid")

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.regplot(
    x = "age",
    y = "fare",
    data=titanic,
    ax=axes[0]
)

sns.regplot(
    x = "age",
    y = "fare",
    data=titanic,
    ax=axes[1],
    fit_reg=False
)


plt.show()
```

```python
# 히스토그램/커널 밀도 그래프

fig, axes = plt.subplot_mosaic([
    ["top_left", "top_right"],
    ["middle_left", "middle_right"],
    ["bottom", "bottom"]
])

sns.histplot(
    x="age",
    data=titanic,
    ax=axes["top_left"]
)

# 2. hue옵션 추가
sns.histplot(
    x="age",
    hue="survived",
    data=titanic,
    ax=axes["top_right"]
)

# dodge 적용
sns.histplot(
    x="age",
    hue="survived",
    data=titanic,
    ax=axes["middle_left"],
    multiple="dodge"
)

# stack
sns.histplot(
    x="age",
    hue="survived",
    multiple="stack",
    data=titanic,
    ax=axes["middle_right"]
)

# multiple = fill
sns.histplot(
    x="age",
    hue="survived",
    multiple="fill",
    data=titanic,
    ax=axes["bottom"]
)

plt.show()
```
